```python
from pathlib import Path

SRC_DIR = Path("/projects/dagr5998/icepack-dirs/input/forcing/sibchuk_combined/JRA55")
N_HEADER = 2
HOURS_PER_YEAR = 8760
WINDOW_YEARS = 3
START_YEAR = 2000
N_WINDOWS = 14

for src in sorted(SRC_DIR.glob("ATM_FORCING_[0-9][0-9][0-9][0-9].txt")):
    lines = src.read_text().splitlines(keepends=True)
    header, data = lines[:N_HEADER], lines[N_HEADER:]
    assert len(data) == 16 * HOURS_PER_YEAR, f"{src.name}: {len(data)} records"

    member = src.stem.split("_")[-1]
    for w in range(N_WINDOWS):
        y0 = START_YEAR + w
        chunk = data[w * HOURS_PER_YEAR : (w + WINDOW_YEARS) * HOURS_PER_YEAR]
        assert len(chunk) == WINDOW_YEARS * HOURS_PER_YEAR
        out = SRC_DIR / f"ATM_FORCING_{member}_{y0}-{y0+2}.txt"
        out.write_text("".join(header + chunk))

    print(f"{src.name} -> {N_WINDOWS} windows")
```
